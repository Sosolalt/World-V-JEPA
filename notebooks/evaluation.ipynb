Thin wrapper around `scripts/evaluate.py`. Run top-to-bottom to reproduce every figure and metric from a checkpoint + dataset.

Edit the paths in the next cell, then `Run All`.

```python
import sys
from pathlib import Path

REPO_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
if str(REPO_ROOT) not in sys.path:
    sys.path.insert(0, str(REPO_ROOT))

CKPT       = REPO_ROOT / 'runs' / '20260517_231533' / 'ckpt_best.pt'
PIXEL_CKPT = REPO_ROOT / 'runs' / 'baseline_pixel' / '20260517_231915' / 'ckpt_best.pt'
DATA       = REPO_ROOT / 'data' / 'sample' / 'sample.npz'
CONFIG     = REPO_ROOT / 'configs' / 'simple.yaml'
OUT        = REPO_ROOT / 'assets' / 'results'
OUT.mkdir(parents=True, exist_ok=True)
print('repo:', REPO_ROOT)
```

```python
from scripts.evaluate import run_evaluation

pixel_arg = str(PIXEL_CKPT) if PIXEL_CKPT.exists() else None
metrics = run_evaluation(
    ckpt_path=str(CKPT),
    data_path=str(DATA),
    config_path=str(CONFIG),
    out_dir=str(OUT),
    pixel_ckpt_path=pixel_arg,
)
```

```python
from IPython.display import Image, display
figs = [
    'latent_pca_trajectory.png',
    'degradation_curve.png',
    'tsne_by_speed.png',
    'training_curves.png',
    'jepa_vs_pixel.png',
]
for f in figs:
    p = OUT / f
    if p.exists():
        print(f)
        display(Image(filename=str(p)))
    else:
        print('missing:', f)
```
